```python
# import libraries
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
%matplotlib inline
import random
```

```python
# reading data
words = open('names.txt', 'r').read().splitlines()
chars = sorted(list(set(''.join(words))))
stoi = {s:i+1 for i,s in enumerate(chars)}
stoi['.'] = 0
itos = {i:s for s,i in stoi.items()}
```

```python
# Build datasets
block_size = 3

def build_dataset(words):
    X, Y = [], []
    
    for w in words:
        context = [0] * block_size
        
        for char in w + '.':
            ix = stoi[char]
            X.append(context)
            Y.append(ix)
            context = context[1:] + [stoi[char]]
    
    X = torch.tensor(X)
    Y = torch.tensor(Y)

    return X, Y

random.seed(42)
random.shuffle(words)
n1 = int(0.8 * len(words))
n2 = int(0.9 * len(words))
Xtr, Ytr = build_dataset(words[:n1])
Xdev, Ydev = build_dataset(words[n1:n2])
Xtest, Ytest = build_dataset(words[n2:])
```

```python
# Initialize parameters
n_embd = 20
n_hidden = 300

g = torch.Generator().manual_seed(2147483647)
C = torch.randn((27, n_embd), generator=g)
W1 = torch.randn((block_size*n_embd, n_hidden), generator=g) * (5/3) / ((block_size*n_embd)**0.5)
b1 = torch.randn(n_hidden, generator=g) * 0.01
W2 = torch.randn((n_hidden, 27), generator=g) * 0.1
b2 = torch.randn(27, generator=g) * 0.01
parameters = [C, W1, b1, W2, b2]
for p in parameters:
    p.requires_grad = True
```

```python
# use for finding optimal learning rate
lre = torch.linspace(-3, 0, 1000)
lrs = 10**lre
lri = []
lossi = []
stepi = []
```

```python
# training
iterations = 100000
batch_size = 100

for i in range(iterations):
    ix = torch.randint(0, Xtr.shape[0], (batch_size,), generator=g)
    
    emb = C[Xtr[ix]]
    embcat = emb.view(emb.shape[0], -1)
    hpreact = embcat @ W1 + b1
    hpreact = (hpreact - hpreact.mean(0, keepdims=True)) / hpreact.std(0, keepdims=True)
    h = torch.tanh(hpreact)
    logits = h @ W2 + b2
    loss = F.cross_entropy(logits, Ytr[ix])
    if i % 1000 == 0:
        print(loss.item())
    for p in parameters:
        p.grad = None
    loss.backward()
    
    lr = 0.1 if i <= 50000 else 0.01
    for p in parameters:
        p.data += -lr * p.grad

    break
    # lri.append(lre[i])
    # lossi.append(loss.item())
    # stepi.append(i)
```

```python
# Calculate loss on training data
emb = C[Xtr]
h = torch.tanh(emb.view(-1, block_size*n_embd) @ W1 + b1)
logits = h @ W2 + b2
loss = F.cross_entropy(logits, Ytr)

loss.item()
```

```python
# Calculate loss on dev data
emb = C[Xdev]
h = torch.tanh(emb.view(-1, block_size*n_embd) @ W1 + b1)
logits = h @ W2 + b2
loss = F.cross_entropy(logits, Ydev)

loss.item()
```

```python
# Get samples (names) from the machine
for _ in range(10):
    out = []
    context = [0] * block_size

    while True:
        emb = C[torch.tensor(context)]
        h = torch.tanh(emb.view(-1, block_size*n_embd) @ W1 + b1)
        logits = h @ W2 + b2
        probs = F.softmax(logits, dim=1)
        ix = torch.multinomial(probs, num_samples=1, generator=g).item()
        context = context[1:] + [ix]
        out.append(itos[ix])
        if ix == 0:
            break

    print(''.join(out))
        
```
